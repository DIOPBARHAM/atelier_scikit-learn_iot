## Partie 2 – Sélection de y (cible) et X (caractéristiques) 
### 1) Définir "etat" comme la cible ou valeur à prédire et "temperature", "humidite", "pression" et "consommation" comme caractéristiques ou variables explicatives 

```python
# Vérification des valeurs manquantes dans la colonne cible 'etat'
nan_etat = df['etat'].isnull().sum()
print(f"Nombre de valeurs manquantes dans la variable cible 'etat' : {nan_etat}")

# Suppression des lignes où la variable cible 'etat' est manquante
if nan_etat > 0:
    df = df.dropna(subset=['etat'])
    print("Lignes avec variable cible manquante supprimées !")

print(f"Dimensions du DataFrame df après nettoyage : {df.shape[0]} lignes.")

#  Définir 'etat' comme la cible ou valeur à prédire et 'temperature', 'humidite', 'pression' et 'consommation' comme caractéristiques
X = df[["temperature", "humidite", "pression", "consommation"]]
y = df["etat"]
```

### 2) Afficher les cinq premières lignes de X et de y


```python
print("--- 5 premières lignes de X (Caractéristiques ou variables explicatives) ---")
display(X.head())
print("\n--- 5 premières lignes de y (Variable cible ou étiquette) ---")
display(y.head())
```

### 3) Quel est le type du problème de machine learning ? 


Il s'agit d'un problème de **classification multiclasse** (qui relève de l'apprentissage supervisé).

* **Pourquoi de l'apprentissage supervisé ?** Car nous disposons de données étiquetées : chaque mesure physique est associée à sa "bonne réponse" stockée dans la variable cible `etat` (OK, ALERTE, ERREUR).
* **Pourquoi de la classification ?** Car la variable à prédire est **qualitative/catégorielle** (l'état prend des modalités textuelles définies : OK, ALERTE, ERREUR), par opposition à un problème de régression où la variable cible est numérique continue (comme prédire un prix ou une température) .
* **Pourquoi multiclasse ?** Car la variable cible comporte **plus de deux classes** (ici 3 classes distinctes), par opposition à une classification binaire qui n'en comporte que deux.

## Partie 3 – Découpage Train/Test 

```python
from sklearn.model_selection import train_test_split

# Découpage en ensembles d'entraînement (80%) et de test (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size=0.2, # 20% des données serviront au test
    random_state=42, # Garantir la reproductibilité
    stratify=y     # Conserver les mêmes proportions de classes
)

print(f"Dimensions de X_train (Entraînement) : {X_train.shape[0]} lignes, {X_train.shape[1]} caractéristiques")
print(f"Dimensions de X_test (Test)          : {X_test.shape[0]} lignes, {X_test.shape[1]} caractéristiques")
```

```python
from collections import Counter

# Vérification de la répartition des classes dans les deux ensembles
print("Répartition des classes dans l'ensemble d'entraînement (y_train) :")
for classe, count in Counter(y_train).items():
    print(f" - {classe:7s} : {count:3d} exemples ({count/len(y_train)*100:.2f}%)")

print("\nRépartition des classes dans l'ensemble de test (y_test) :")
for classe, count in Counter(y_test).items():
    print(f" - {classe:7s} : {count:3d} exemples ({count/len(y_test)*100:.2f}%)")
```

## Partie 4 – Gestion des valeurs manquantes 
### 1) Vérifier l’existence de valeurs manquantes 

```python
print("--- Valeurs manquantes dans X_train ---")
print(X_train.isnull().sum())
print("\n--- Valeurs manquantes dans X_test ---")
print(X_test.isnull().sum())
```

### 2) Sélectionner SimpleImputer avec la médiane 

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="median")
```

### 3) Qu’est ce qui justifie le choix de la médiane ? 

Le choix de la **médiane** (strategy="median") pour remplacer les valeurs manquantes est justifié par sa **robustesse face aux valeurs aberrantes ou extrêmes** (outliers).
Contrairement à la moyenne (qui est fortement influencée par des anomalies, par exemple si un capteur de température défectueux renvoie temporairement 150°C ou si un pic anormal est mesuré), la médiane représente la valeur centrale du jeu de données et reste stable même en présence d'anomalies de mesure. C'est l'imputation par défaut recommandée en milieu industriel et IoT pour préserver la cohérence des distributions.


### 4) Trouver les paramètres (médianes) de l’imputeur sur X_train 

```python
#On calcule la statistique sur X_train uniquement, et on l'applique sur X_train et X_test.
imputer.fit(X_train)

print("Médianes calculées par variable sur l'ensemble d'entraînement :")
for col, med in zip(X_train.columns, imputer.statistics_):
    print(f" - {col:12s} : {med:.2f}")
```

### 5) Déterminer X_train_imputed et X_test_imputed, les transformés de X_train et X_test

```python
import numpy as np

X_train_imputed = imputer.transform(X_train)
X_test_imputed = imputer.transform(X_test)
# Vérification finale de l'absence de valeurs manquantes
print(f"Valeurs manquantes restantes dans X_train_imputed : {np.isnan(X_train_imputed).sum()}")
print(f"Valeurs manquantes restantes dans X_test_imputed  : {np.isnan(X_test_imputed).sum()}")
```

## Partie 5 – Mise à l'échelle 
### 1) Sélectionner StandardScaler pour mettre à l’échelle les transformés de l’imputation 

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

### 2) Qu’est ce qui justifie la standardisation ? 

Certains modèles de machine learning, en particulier le modèle des **k plus proches voisins (KNN)** que nous allons utiliser, reposent sur le calcul de la **distance géométrique** (distance euclidienne) entre les points pour effectuer une classification.

Si l'on n'effectue pas de mise à l'échelle, les caractéristiques qui ont de grandes valeurs numériques vont totalement écraser et dominer le calcul des distances par rapport aux variables à plus petite échelle. Le modèle croira alors à tort que la pression est beaucoup plus importante que la température.

La standardisation (StandardScaler) résout ce problème en recentrant et réduisant chaque variable pour qu'elle ait une **moyenne proche de 0** et un **écart-type proche de 1**, mettant ainsi toutes les variables sur un pied d'égalité.


### 3) Trouver les paramètres (moyennes et écart-types) du scaleur sur X_train_imputed

```python
#Toujours calculé uniquement sur le jeu d'entraînement pour éviter le data leakage
scaler.fit(X_train_imputed)

print("Paramètres appris par le StandardScaler (Moyennes et Écarts-types) :")
for col, mean, std in zip(X_train.columns, scaler.mean_, np.sqrt(scaler.var_)):
    print(f" - {col:12s} : Moyenne = {mean:7.2f} | Écart-type = {std:6.2f}")
```

### 4) Déterminer X_train_scaled et X_test_scaled, les transformés de X_train_imputed et X_test_imputed

```python
X_train_scaled = scaler.transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

print("--- Aperçu de la première ligne avant/après standardisation ---")
print("Brut (Imputé) :", X_train_imputed[0])
print("Standardisé    :", X_train_scaled[0])
```

## Partie 6 – Entrainement et prédiction d’un modèle 
### 1) Sélectionner le modèle KNN (k plus proches voisins) avec 5, le nombre de voisins à prendre en compte 

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=5)
```

### 2) Entrainer le modèle


```python
model_knn.fit(X_train_scaled, y_train)
print("Modèle KNN entraîné avec succès !")
```

### 3) Déterminer y_pred, la prédiction du modèle avec l’ensemble de test

```python
y_pred = model_knn.predict(X_test_scaled)
```

### 4) Afficher quelques prédictions (les 10 premières)

```python
print("Les 10 premières prédictions du modèle (y_pred) :")
print(y_pred[:10])

print("\nLes 10 vraies valeurs réelles correspondantes (y_test) :")
print(y_test[:10].tolist())
```

### 5) Comparer ces prédictions avec les vraies valeurs

```python
comparaison_df = pd.DataFrame({
    'Température (°C)': X_test['temperature'].values,
    'Humidité (%)': X_test['humidite'].values,
    'Vraie valeur (y_test)': y_test.values,
    'Prédiction (y_pred)': y_pred,
    'Correct': y_test.values == y_pred
})

print("--- Les 10 premières lignes du jeu de test comparées aux prédictions ---")
display(comparaison_df.head(10))
```

## Partie 7 – Evaluation du modèle 
### 1) Evaluer le modèle avec Accuracy puis afficher le résultat 

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print(f"Exactitude globale (Accuracy) du modèle KNN : {accuracy * 100:.2f}%")
```

### 2) Pourquoi Accuracy ne suffit pas toujours pour évaluer ?

L'Accuracy (taux de bonne classification globale) est une métrique très trompeuse lorsque **les classes sont déséquilibrées (imbalanced dataset)**, ce qui est particulièrement le cas ici.

Dans notre jeu de données, la classe "OK" représente environ 94% des mesures (113 sur 120 dans le jeu de test). Si nous avions un modèle très bête qui prédisait systématiquement "OK" pour toutes les lignes sans analyser les variables, il obtiendrait une exactitude (accuracy) de 94.1% ! Pourtant, un tel modèle serait totalement inutile car il ne parviendrait jamais à détecter les alertes de pannes ("ALERTE") ou les pannes franches ("ERREUR"), ce qui est le but même de notre projet de maintenance préventive.

Il est donc capital de regarder en détail comment le modèle se comporte sur chaque classe à travers la matrice de confusion et le rapport de classification .

### 3) Déterminer, afficher et commenter la matrice de confusion

```python
from sklearn.metrics import confusion_matrix

classes = ['OK', 'ALERTE', 'ERREUR']
matrice = confusion_matrix(y_test, y_pred, labels=classes)
# Affichage sous forme de DataFrame pour une lecture simple et intuitive
matrice_df = pd.DataFrame(
    matrice,
    index=[f"Réel {c}" for c in classes],
    columns=[f"Prédit {c}" for c in classes]
)
print("Matrice de confusion :")
display(matrice_df)
```

### Commentaire de la matrice de confusion :
* **Classe OK (Normal) :** Le modèle a parfaitement identifié les **113 cas de fonctionnement normal** (aucun faux négatif, aucun capteur normal classifié en alerte ou erreur).
* **Classe ALERTE :** Sur **6 cas réellement en alerte**, le modèle en a correctement détecté **2**, mais en a confondu **4 avec le fonctionnement OK** (ce sont des faux négatifs d'anomalie, très embêtants car on va passer à côté de dérives).
* **Classe ERREUR :** Sur l'**unique cas réel de panne** dans notre ensemble de test, le modèle l'a classifié par erreur en **ALERTE** et aucun en ERREUR. Le KNN s'est un peu trompé mais reste proche de l'état d'anomalie.

Le modèle est excellent pour détecter le fonctionnement normal (OK), mais peine encore sur les anomalies à cause de la faible quantité d'ALERTES et d'ERREURS dans les données d'apprentissage (déséquilibre des classes).


### 4) Visualiser cette matrice avec Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.heatmap(matrice_df, annot=True, cmap="Reds", fmt="d", cbar=False)
plt.title("Visualisation graphique de la Matrice de Confusion")
plt.ylabel("Vraies classes")
plt.xlabel("Classes prédites")
plt.tight_layout()
plt.show()
```

### 5) Afficher et commenter le rapport de classification

```python
from sklearn.metrics import classification_report

print("Rapport de classification détaillé par classe :")
print(classification_report(y_test, y_pred, labels=classes, zero_division=0))
```

### Commentaire du rapport de classification :
* **Classe OK :** Précision de 0.97 et Rappel de 1.00. Les scores sont excellents (F1-score de 0.98), car le fonctionnement normal est très simple à reconnaître et possède de nombreuses données d'apprentissage.
* **Classe ALERTE :** Précision de 0.67 (lorsque le modèle prédit ALERTE, il a raison dans 67% des cas), mais un Rappel faible de 0.33 (le modèle ne capture que 33% des vraies alertes de notre jeu de test).
* **Classe ERREUR :** Précision et Rappel de 0.00 en raison du manque flagrant de données (1 seul exemple de test). Le modèle n'a pas pu s'entraîner correctement pour cette classe.

### 6) Dans quel cas de problème de classification est-il plus indiqué d’évaluer avec :

* **La Précision (Precision) :** Indiquée lorsque les **faux positifs (FP) sont extrêmement coûteux ou gênants**.
  * *Exemple :* Détection de spams. Classer à tort un e-mail professionnel crucial comme spam (faux positif) est dramatique, car l'utilisateur peut rater une opportunité majeure. On veut une précision proche de 100% pour être certain de ne pas filtrer de vrais mails.
* **Le Rappel (Recall) :** Indiqué lorsque les **faux négatifs (FN) sont extrêmement dangereux ou critiques**.
  * *Exemple :* Diagnostic médical de maladies graves ou maintenance d'organes vitaux sur une machine. Si on ne détecte pas une maladie grave (faux négatif), le patient risque de mourir. En maintenance industrielle, rater un capteur en ERREUR (faux négatif) peut conduire à une panne catastrophique et coûteuse de toute la ligne de production. On préfère avoir quelques fausses alertes à vérifier plutôt que de rater une vraie panne.
* **Le F1-Score :** Indiqué lorsqu'on recherche un **bon compromis équilibré entre précision et rappel sur des classes déséquilibrées** . C'est la moyenne harmonique des deux. Si le F1-score est élevé, cela garantit que notre modèle est à la fois fiable dans ses prédictions (bonne précision) et efficace pour détecter les anomalies (bon rappel).
* **L'Accuracy (Exactitude) :** Indiquée uniquement lorsque **les classes du jeu de données sont parfaitement équilibrées** (même proportion de OK, ALERTE et ERREUR) et que toutes les erreurs de classification ont le même impact économique ou sécuritaire.
